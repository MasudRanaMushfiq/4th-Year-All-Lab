```python
import heapq

def a_star(graph, heuristics, start, goal):
    open_set = []
    heapq.heappush(open_set, (heuristics[start], 0, start, None))
    
    came_from = {}
    g_scores = {start: 0}
    
    while open_set:
        current_f, current_g, current_node, parent = heapq.heappop(open_set)
        
        if current_node == goal:
           
            path = []
            while current_node is not None:
                path.append(current_node)
                current_node = came_from.get(current_node)
            return path[::-1]  
        
        if current_g > g_scores.get(current_node, float('inf')):
            continue
        
        for neighbor, cost in graph[current_node]:
            tentative_g = current_g + cost
           
            if tentative_g < g_scores.get(neighbor, float('inf')):
                came_from[neighbor] = current_node
                g_scores[neighbor] = tentative_g
                f = tentative_g + heuristics[neighbor]
                heapq.heappush(open_set, (f, tentative_g, neighbor, current_node))
    
    return None  

# Given graph and heuristics
graph = {
    'A': [('B', 3), ('D', 2)],
    'B': [('C', 1), ('E', 4), ('F', 2)],
    'C': [('F', 5)],
    'D': [('E', 1)],
    'E': [('F', 3)],
    'F': []
}

heuristics = {
    'A': 6,
    'B': 4,
    'C': 2,
    'D': 5,
    'E': 3,
    'F': 0
}

# Perform A* search from 'A' to 'F'
path = a_star(graph, heuristics, 'A', 'F')
print("Path found by A*:", path)
```

```python
import heapq

def manhattan_distance(state):
    distance = 0
    for i in range(3):
        for j in range(3):
            num = state[i][j]
            if num != 0:
                target_row = (num - 1) // 3
                target_col = (num - 1) % 3
                distance += abs(i - target_row) + abs(j - target_col)
    return distance

def misplaced_tiles(state):
    goal = ((1, 2, 3), (8, 0, 4), (7, 6, 5))
    count = 0
    for i in range(3):
        for j in range(3):
            if state[i][j] != goal[i][j] and state[i][j] != 0:
                count += 1
    return count

def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return (i, j)
    return None

def generate_moves(state):
    moves = []
    blank_i, blank_j = find_blank(state)
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]  # Up, Down, Left, Right

    for di, dj in directions:
        new_i, new_j = blank_i + di, blank_j + dj
        if 0 <= new_i < 3 and 0 <= new_j < 3:
            # Convert tuple to list to perform swap
            state_list = [list(row) for row in state]
            state_list[blank_i][blank_j], state_list[new_i][new_j] = state_list[new_i][new_j], state_list[blank_i][blank_j]
            new_state = tuple(tuple(row) for row in state_list)
            moves.append(new_state)
    return moves

def a_star(start, goal, heuristic):
    open_set = []
    heapq.heappush(open_set, (heuristic(start), 0, start))
    
    came_from = {}
    g_scores = {start: 0}
    
    while open_set:
        current_f, current_g, current_state = heapq.heappop(open_set)
        
        if current_state == goal:
            # Reconstruct path
            path = []
            while current_state in came_from:
                path.append(current_state)
                current_state = came_from[current_state]
            path.append(start)
            return path[::-1]  # Reverse to show start to goal
        
        if current_g > g_scores.get(current_state, float('inf')):
            continue
        
        for neighbor in generate_moves(current_state):
            tentative_g = current_g + 1
            if tentative_g < g_scores.get(neighbor, float('inf')):
                came_from[neighbor] = current_state
                g_scores[neighbor] = tentative_g
                f = tentative_g + heuristic(neighbor)
                heapq.heappush(open_set, (f, tentative_g, neighbor))
    
    return None  # No solution exists

# Example usage:
if __name__ == "__main__":
    start = (
        (2, 8, 3),
        (1, 6, 4),
        (7, 0, 5)
    )
    goal = (
        (1, 2, 3),
        (8, 0, 4),
        (7, 6, 5)
    )

    print("Solving using Manhattan Distance heuristic:")
    path_manhattan = a_star(start, goal, manhattan_distance)
    if path_manhattan:
        print(f"Path length: {len(path_manhattan) - 1} moves")
        for state in path_manhattan:
            print("Heuristics State")
            for row in state:
                print(" ".join(str(num) for num in row))
    else:
        print("No solution found!")

    print("\nSolving using Misplaced Tiles heuristic:")
    path_misplaced = a_star(start, goal, misplaced_tiles)
    if path_misplaced:
        print(f"Path length: {len(path_misplaced) - 1} moves")
        for state in path_misplaced:
            print("-----")
            for row in state:
                print(" ".join(str(num) for num in row))
    else:
        print("No solution found!")
```
